```python
import pandas as pd
```

```python
vacancies = pd.read_csv("vacancies_rows.csv")
skills = pd.read_csv("skills_rows.csv")
vacancy_skills = pd.read_csv("vacancy_skills_rows.csv")
```

```python
print("Vacancies:", vacancies.shape)
print("Skills:", skills.shape)
print("Vacancy-Skills:", vacancy_skills.shape)
```

```python
print("\nVacancies columns:")
print(vacancies.columns.tolist())
```

```python
print("\nVacancy-Skills columns:")
print(vacancy_skills.columns.tolist())
```

```python
print("\nSkills columns:")
print(skills.columns.tolist())
```

```python
print("Missing values in Vacancies:")
print(vacancies.isnull().sum())

print("\nMissing values in Skills:")
print(skills.isnull().sum())

print("\nMissing values in Vacancy-Skills:")
print(vacancy_skills.isnull().sum())
```

```python
print("Duplicate rows in Vacancies:",
      vacancies.duplicated().sum())

print("Duplicate rows in Skills:",
      skills.duplicated().sum())

print("Duplicate rows in Vacancy-Skills:",
      vacancy_skills.duplicated().sum())
```

```python
print("Duplicate Vacancy IDs:",
      vacancies["id"].duplicated().sum())

print("Duplicate Skill IDs:",
      skills["id"].duplicated().sum())
```

```python
print(vacancies["experience_years"].describe())

print("\nUnique experience levels:")
print(vacancies["experience_level"].value_counts(dropna=False))
```

```python
# Check whether all vacancy IDs in the bridge table exist
invalid_vacancies = ~vacancy_skills["vacancy_id"].isin(vacancies["id"])

# Check whether all skill IDs in the bridge table exist
invalid_skills = ~vacancy_skills["skill_id"].isin(skills["id"])

print("Invalid vacancy IDs:", invalid_vacancies.sum())
print("Invalid skill IDs:", invalid_skills.sum())
```

```python
#clean copies of data 
vacancies_clean = vacancies.copy()
skills_clean = skills.copy()
vacancy_skills_clean = vacancy_skills.copy()

print("Clean copies created successfully.")
```

```python
vacancies_clean["created_at"] = pd.to_datetime(
    vacancies_clean["created_at"],
    errors="coerce"
)

vacancies_clean["published_at"] = pd.to_datetime(
    vacancies_clean["published_at"],
    errors="coerce"
)

print(vacancies_clean[["created_at", "published_at"]].dtypes)
```

```python
#remove unncessary spaces 
text_columns = [
    "title",
    "company",
    "location",
    "experience_level"
]

for col in text_columns:
    vacancies_clean[col] = vacancies_clean[col].astype("string").str.strip()
```

```python
#clean job description
vacancies_clean["description_raw"] = (
    vacancies_clean["description_raw"]
    .astype("string")
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)
```

```python
print(vacancies_clean.info())
```

```python
print(vacancies_clean.head(3))
```

```python
print("Original created_at:")
print(vacancies["created_at"].head(20).to_string())

print("\nCleaned created_at:")
print(vacancies_clean["created_at"].head(20).to_string())
```

```python
vacancies_clean["created_at"] = pd.to_datetime(
    vacancies["created_at"],
    errors="coerce",
    utc=True
)

print(vacancies_clean["created_at"].notna().sum())
```

```python
print(vacancies_clean["created_at"].head())
```

```python
vacancies_clean["created_date"] = (
    vacancies_clean["created_at"].dt.date
)

vacancies_clean["created_year"] = (
    vacancies_clean["created_at"].dt.year
)

vacancies_clean["created_month"] = (
    vacancies_clean["created_at"].dt.month
)
```

```python
print(vacancies_clean[[
    "created_at",
    "created_date",
    "created_year",
    "created_month"
]].head())

print("\nMissing created_at:",
      vacancies_clean["created_at"].isna().sum())
```

```python
print(vacancies["created_at"].dropna().astype(str).head(50).to_list())
```

```python
vacancies_clean["created_at"] = pd.to_datetime(
    vacancies["created_at"],
    errors="coerce",
    utc=True,
    format="mixed"
)

print("Valid created_at:", vacancies_clean["created_at"].notna().sum())
print("Missing created_at:", vacancies_clean["created_at"].isna().sum())
```

```python
vacancies_clean["created_date"] = vacancies_clean["created_at"].dt.date
vacancies_clean["created_year"] = vacancies_clean["created_at"].dt.year
vacancies_clean["created_month"] = vacancies_clean["created_at"].dt.month

print(vacancies_clean[[
    "created_at",
    "created_date",
    "created_year",
    "created_month"
]].head())
```

```python
print("\nMissing created_at:",
      vacancies_clean["created_at"].isna().sum())
```

```python
vacancies_clean["description_available"] = (
    vacancies_clean["description_raw"].notna()
).astype(int)

print(
    vacancies_clean["description_available"].value_counts()
) 
#1-JD available 0-not available
```

```python
vacancies_clean["experience_years_status"] = (
    vacancies_clean["experience_years"]
    .notna()
    .map({
        True: "Specified",
        False: "Not specified"
    })
)

print(vacancies_clean["experience_years_status"].value_counts())
```

```python
print("Missing published_at:",
      vacancies_clean["published_at"].isna().sum())

print("Earliest published date:",
      vacancies_clean["published_at"].min())

print("Latest published date:",
      vacancies_clean["published_at"].max())
```

```python
vacancies_clean["location_clean"] = (
    vacancies_clean["location"]
    .fillna("Not specified")
    .str.strip()
)

print("Missing location_clean:",
      vacancies_clean["location_clean"].isna().sum())

print("\nTop 15 locations:")
print(
    vacancies_clean["location_clean"]
    .value_counts()
    .head(15)
)
```

```python
vacancies_clean["experience_level_clean"] = (
    vacancies_clean["experience_level"]
    .fillna("Not specified")
    .str.strip()
)

print("Missing experience_level_clean:",
      vacancies_clean["experience_level_clean"].isna().sum())

print("\nExperience levels:")
print(
    vacancies_clean["experience_level_clean"]
    .value_counts()
)
```

```python
print("========== JOBPULSE DATA QUALITY CHECK ==========")

print("\nRows:", len(vacancies_clean))
print("Columns:", len(vacancies_clean.columns))

print("\nDuplicate rows:",
      vacancies_clean.duplicated().sum())

print("Duplicate vacancy IDs:",
      vacancies_clean["id"].duplicated().sum())

print("\nMissing values:")
print(
    vacancies_clean.isna().sum()
)
```

```python
print("========== SKILLS DATA QUALITY CHECK ==========")

print("Rows:", len(skills_clean))
print("Columns:", len(skills_clean.columns))

print("\nMissing values:")
print(skills_clean.isna().sum())

print("\nDuplicate rows:",
      skills_clean.duplicated().sum())

print("Duplicate skill IDs:",
      skills_clean["id"].duplicated().sum())

print("\nDuplicate skill names:",
      skills_clean["name"].duplicated().sum())
```

```python
skills_clean["skill_name"] = (
    skills_clean["name"]
    .astype("string")
    .str.strip()
)

print("Missing skill names:",
      skills_clean["skill_name"].isna().sum())

print("\nSample cleaned skills:")
print(
    skills_clean[["id", "name", "skill_name"]].head(20)
)
```

```python
print("========== VACANCY-SKILLS QUALITY CHECK ==========")

print("Rows:", len(vacancy_skills_clean))
print("Columns:", len(vacancy_skills_clean.columns))

print("\nMissing values:")
print(vacancy_skills_clean.isna().sum())

print("\nDuplicate rows:",
      vacancy_skills_clean.duplicated().sum())

print("\nUnique vacancies with skills:",
      vacancy_skills_clean["vacancy_id"].nunique())

print("Unique skills used:",
      vacancy_skills_clean["skill_id"].nunique())
```

```python
skills_per_vacancy = (
    vacancy_skills_clean
    .groupby("vacancy_id")["skill_id"]
    .nunique()
)

print("Average skills per vacancy:",
      skills_per_vacancy.mean())

print("Minimum skills:",
      skills_per_vacancy.min())

print("Maximum skills:",
      skills_per_vacancy.max())

print("\nSkills per vacancy distribution:")
print(skills_per_vacancy.describe())
```

```python
vacancies_clean["skill_count"] = (
    vacancies_clean["id"]
    .map(skills_per_vacancy)
    .fillna(0)
    .astype(int)
)

print(
    vacancies_clean[
        ["id", "title", "skill_count"]
    ].head(10)
)
```

```python
no_skill_jobs = vacancies_clean[
    vacancies_clean["skill_count"] == 0
]

print("Vacancies with no linked skills:",
      len(no_skill_jobs))

print("\nSample:")
print(
    no_skill_jobs[
        ["id", "title", "company", "description_available"]
    ].head(20)
)
```

```python
vacancies_clean["skill_mapping_status"] = (
    vacancies_clean["skill_count"]
    .gt(0)
    .map({
        True: "Skills mapped",
        False: "No skills mapped"
    })
)

print(
    vacancies_clean["skill_mapping_status"]
    .value_counts()
)
```

```python
skill_demand = (
    vacancy_skills_clean
    .merge(
        skills_clean[["id", "skill_name"]],
        left_on="skill_id",
        right_on="id",
        how="left"
    )
)

print(skill_demand.head(10))
```

```python
print("\nRows:", len(skill_demand))

print(
    "\nMissing skill names after merge:",
    skill_demand["skill_name"].isna().sum()
)
```

```python
top_skills = (
    skill_demand
    .groupby("skill_name")["vacancy_id"]
    .nunique()
    .sort_values(ascending=False)
    .reset_index(name="vacancy_count")
)

print(top_skills.head(20))
```

```python
excel_variants = skills_clean[
    skills_clean["skill_name"].str.contains(
        "excel", case=False, na=False
    )
]

print(
    excel_variants[["id", "skill_name"]]
)
```

```python
skills_clean["skill_name_standard"] = (
    skills_clean["skill_name"]
    .replace({
        "Microsoft Excel": "Excel",
        "MS Excel": "Excel"
    })
)

print(
    skills_clean[
        skills_clean["skill_name"].str.contains(
            "excel", case=False, na=False
        )
    ][["skill_name", "skill_name_standard"]]
)
```

```python
skill_demand = (
    vacancy_skills_clean
    .merge(
        skills_clean[["id", "skill_name_standard"]],
        left_on="skill_id",
        right_on="id",
        how="left"
    )
)

top_skills = (
    skill_demand
    .groupby("skill_name_standard")["vacancy_id"]
    .nunique()
    .sort_values(ascending=False)
    .reset_index(name="vacancy_count")
)

print(top_skills.head(20))
```

```python
common_variants = skills_clean[
    skills_clean["skill_name"].str.contains(
        r"power\s*bi|powerbi|tableau|python|sql",
        case=False,
        na=False,
        regex=True
    )
]

print(
    common_variants[["id", "skill_name"]]
    .sort_values("skill_name")
    .to_string(index=False)
)
```

```python
skills_clean["skill_name_standard"] = (
    skills_clean["skill_name"]
    .replace({
        "Microsoft Power BI": "Power BI",
        "Microsoft PowerBI": "Power BI",
        "PowerBI": "Power BI"
    })
)

print(
    skills_clean[
        skills_clean["skill_name"].str.contains(
            "power.?bi",
            case=False,
            na=False
        )
    ][["skill_name", "skill_name_standard"]]
)
```

```python
skill_demand = (
    vacancy_skills_clean
    .merge(
        skills_clean[["id", "skill_name_standard"]],
        left_on="skill_id",
        right_on="id",
        how="left"
    )
)

top_skills = (
    skill_demand
    .groupby("skill_name_standard")["vacancy_id"]
    .nunique()
    .sort_values(ascending=False)
    .reset_index(name="vacancy_count")
)

print(top_skills.head(20))
```

```python
vacancies_clean.to_csv(
    "vacancies_clean.csv",
    index=False
)

skills_clean.to_csv(
    "skills_clean.csv",
    index=False
)

vacancy_skills_clean.to_csv(
    "vacancy_skills_clean.csv",
    index=False
)

print("Cleaned datasets saved successfully!")
```
